```python
df["FareGroup"] = pd.qcut(
    df["Fare"],
    q=4,
    labels=["Low", "Medium-Low", "Medium-High", "High"]
)

survival_by_fare = df.groupby(
    "FareGroup",
    observed=True
)["Survived"].mean() * 100

print(survival_by_fare)
```

```python
survival_by_fare.plot(
    kind="bar",
    color="#8B72C7",
    edgecolor="black"
)

plt.title("Survival Rate by Fare Group")
plt.xlabel("Fare Group")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.show()
```

```python

survival_by_class_fare = df.pivot_table(
    index="Pclass",
    columns="FareGroup",
    values="Survived",
    aggfunc="mean",
    observed=True
) * 100

print(survival_by_class_fare.round(1))
```

```python

passenger_counts = df.pivot_table(
    index="Pclass",
    columns="FareGroup",
    values="Survived",
    aggfunc="count",
    observed=True
)

print(passenger_counts)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.heatmap(
    survival_by_class_fare,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu",
    vmin=0,
    vmax=100,
    linewidths=0.5
)

plt.title("Survival Rate by Passenger Class and Fare Group")
plt.xlabel("Fare Group")
plt.ylabel("Passenger Class")
plt.show()
```
